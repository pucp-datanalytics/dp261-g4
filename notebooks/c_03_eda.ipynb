# Sprint 1 - EDA

Analiza nulos, duplicados, descriptivas, distribucion del target y visualizaciones basicas.

Este notebook es un entregable academico. La logica principal vive en scripts reproducibles `src/c_*.py` y aqui se documenta/ejecuta el flujo con rutas relativas.

```python
from pathlib import Path
import sys
ROOT = Path.cwd()
sys.path.insert(0, str(ROOT / 'src'))

import matplotlib.pyplot as plt
from c_preprocessing import load_raw_data, TARGET_COL
df = load_raw_data()
print('nulos:', df.isna().sum().sum())
print('duplicados:', df.duplicated().sum())
print(df[TARGET_COL].value_counts())
print(df.describe(include='all').transpose().head(20))
df[TARGET_COL].value_counts().plot(kind='bar', title='Revenue')
plt.show()
```

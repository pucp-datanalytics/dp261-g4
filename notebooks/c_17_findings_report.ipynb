# Sprint 5 - Hallazgos ejecutivos

Este notebook sintetiza conclusiones tecnicas y economicas para la presentacion final.

```python
import pandas as pd

comparison = pd.read_csv('models/c_model_economic_comparison.csv')
baseline = pd.read_csv('models/c_baseline_comparison_usd.csv')
xgb = comparison[comparison['model'].eq('XGBoost')].iloc[0]
print('Modelo recomendado:', xgb['model'])
print('Valor economico estimado USD:', round(xgb['net_value_usd'], 2))
print('ROI estimado %:', round(xgb['roi_pct'], 2))
print('Visitantes impactados:', int(xgb['visitors_impacted']))
```

## Insights principales

1. XGBoost es el modelo recomendado y queda alineado con el modelo final del proyecto.
2. El analisis economico permite comparar modelos por valor neto estimado, no solo por F1-score.
3. El dashboard ya no es estatico: permite cargar nuevos datasets.
4. Si el dataset cargado tiene `Revenue`, se calculan metricas reales y valor economico observado.
5. Si el dataset no tiene `Revenue`, se generan predicciones y valor potencial estimado.
6. El modelo no se reentrena desde el dashboard.

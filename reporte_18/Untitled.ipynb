```python
import subprocess

print("=== BUSCANDO CSV EN RAMA DE PERLA ===\n")

# Ver qué archivos hay en models/
result = subprocess.run(
    ["git", "ls-tree", "origin/feat/PB-16-sprint5-business-value", "models/", "-r", "--name-only"],
    capture_output=True, text=True
)

print("Archivos en carpeta models/ de Perla:")
for archivo in result.stdout.split('\n'):
    if archivo:
        print(f"  - {archivo}")

# Buscar específicamente los CSV
print("\n=== BUSCANDO CSV BUSINESS VALUE ===")
csvs = ["business_value_threshold_search.csv", 
        "business_value_recommendations.csv", 
        "business_value_gain_lift_curve.csv"]

for csv in csvs:
    result = subprocess.run(
        ["git", "ls-tree", "origin/feat/PB-16-sprint5-business-value", f"models/{csv}"],
        capture_output=True, text=True
    )
    if result.stdout:
        print(f"✅ {csv} - EXISTE")
    else:
        print(f"❌ {csv} - NO EXISTE")
```
